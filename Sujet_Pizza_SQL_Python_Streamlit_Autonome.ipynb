# Projet Data Analyst — Plato’s Pizza

**Sujet ouvert : SQL · Python · Streamlit**  

Vous intervenez comme équipe Data Analyst pour **Plato’s Pizza**, une pizzeria disposant de données historiques de commandes et de produits. La direction souhaite disposer d’un outil qui lui permette de mieux comprendre son activité et d’appuyer ses décisions commerciales.

**Votre mission : concevoir une analyse de l’activité et la rendre accessible dans une application Streamlit.**


## Mission

À partir des données fournies, vous devez :

- construire une **base de données relationnelle SQL** exploitable ;
- utiliser **SQL** pour répondre à des questions d’analyse commerciale ;
- utiliser **Python** pour traiter, contrôler, analyser et/ou visualiser les résultats ;
- concevoir une **application Streamlit interactive** utile à la direction ;
- présenter vos principaux constats et proposer des **recommandations justifiées par les données**.

**Le cadrage analytique vous appartient.** Définissez les questions métier qui vous paraissent pertinentes, les indicateurs à suivre, le niveau de détail et les fonctionnalités de l’application. Vous choisissez librement le SGBD (SQLite, PostgreSQL, MySQL…), les bibliothèques Python et l’organisation du projet.

Le projet doit démontrer une réelle utilisation des trois composantes : **SQL, Python et Streamlit**. Une application qui lit uniquement les CSV sans exploiter la base SQL ne répond pas à l’objectif de révision.


## Données

Le jeu de données **Pizza Place Sales** comprend quatre fichiers CSV liés :

| Fichier | Contenu |
|---|---|
| `orders.csv` | Identifiants, dates et heures des commandes |
| `order_details.csv` | Lignes de commande, pizzas et quantités |
| `pizzas.csv` | Références, tailles et prix des pizzas |
| `pizza_types.csv` | Types de pizzas, catégories et ingrédients |

**Téléchargement :**

- [Jeu de données d’origine — Maven Analytics](https://mavenanalytics.io/data-playground/pizza-place-sales)

La présence de fichiers CSV ne dispense pas de la création et de l’utilisation d’une base de données SQL.


## Livrables attendus — vendredi matin

1. **Base SQL :** structure relationnelle et données chargées ; scripts ou méthode permettant de reproduire la création.
2. **Analyse :** requêtes SQL et code Python utilisés, organisés et compréhensibles.
3. **Application Streamlit :** interface interactive permettant de consulter les résultats de l’analyse à partir de la base SQL.
4. **Restitution :** courte démonstration de l’application, principaux enseignements, limites et recommandations.
5. **README :** instructions pour installer les dépendances, préparer la base et lancer l’application.

Vous êtes libres de la répartition du travail, de la structure du dépôt, du nombre de pages, des graphiques et des indicateurs. L’essentiel est de justifier vos choix et de pouvoir expliquer votre code.


## Critères d’évaluation

- **SQL :** cohérence du modèle, utilisation pertinente des jointures et des agrégations, fiabilité des résultats.
- **Python :** qualité du traitement et de l’analyse, lisibilité et organisation du code.
- **Streamlit :** application fonctionnelle, filtres ou interactions utiles, clarté de la présentation.
- **Raisonnement Data Analyst :** pertinence des questions métier, interprétation des résultats, recommandations fondées sur les données.
- **Autonomie :** capacité à expliquer et défendre les choix techniques et analytiques.

**Aucun modèle de solution, aucune requête imposée et aucun squelette d’application ne sont fournis.**
